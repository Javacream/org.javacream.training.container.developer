# Jenkins Build-Umgebung mit Docker Compose
## Controller + 2 Agents (Inbound / WebSocket)

**Ziel dieses Notebooks**  
Am Ende dieses Tutorials kannst du eine lokale Jenkins-Build-Umgebung mit
- **1 Jenkins Controller**
- **2 Jenkins Agents**
- Verbindung über **Inbound Agent + WebSocket**
aufsetzen und betreiben.

Die Umgebung eignet sich ideal für **CI/CD-Experimente, Schulungen und lokale Tests**.

## 1. Architektur-Überblick

**Komponenten**
- Jenkins Controller  
  - Web UI
  - Job-Definitionen
  - Koordination der Builds
- Jenkins Agents (2x)
  - führen Builds aus
  - verbinden sich aktiv zum Controller

**Warum Inbound Agents (WebSocket)?**
- kein SSH notwendig
- funktioniert gut in Docker-Netzwerken
- weniger Port-Freigaben

## 2. Voraussetzungen

- Docker Engine oder Docker Desktop
- Docker Compose (Plugin oder Standalone)
- Browser (für Jenkins UI)

Prüfe lokal:
```bash
docker --version
docker compose version
```

## 3. docker-compose.yml

Diese Compose-Datei startet:
- `jenkins` → Controller
- `agent1` → Build-Agent 1
- `agent2` → Build-Agent 2

## 4. .env Datei (Secrets für Agents)

Die Agent-Secrets kommen **aus der Jenkins UI** und werden **nicht** im Compose-File gespeichert.

Lege eine Datei `.env` im selben Ordner an:

## 5. Jenkins starten

```bash
docker compose up -d
```

Danach:
- Jenkins UI: http://localhost:8080

## 6. Initiales Jenkins Setup

1. Initiales Admin-Passwort holen:
```bash
docker exec -it jenkins-controller   cat /var/jenkins_home/secrets/initialAdminPassword
```

2. Browser öffnen → Passwort eingeben
3. „Install suggested plugins“
4. Admin-User anlegen

## 7. Agents in Jenkins anlegen

Für **jeden Agenten**:

1. *Manage Jenkins → Nodes → New Node*
2. Name: `agent-1` bzw. `agent-2`
3. Type: **Permanent Agent**
4. Launch method:
   - **Launch agent by connecting it to the controller**
5. Secret kopieren
6. Secret in `.env` eintragen

## 8. Agents verbinden

Nachdem die Secrets eingetragen sind:

```bash
docker compose up -d
```

Jetzt sollten beide Agents im Jenkins UI als **online** erscheinen.

## 10. Typische Erweiterungen

- Docker-in-Docker Agenten
- Separate Agent-Typen (Java, Node, Python)
- Reverse Proxy (HTTPS)
- Backup von `jenkins_home`
